# EndSurvGAN — Quick Research Demo

This notebook demonstrates the repository structure and a minimal forward pass through the EndSurvGAN generator/discriminator. It does not reproduce the full conference experiment without the benchmark datasets and complete preprocessing pipeline.

```python
import sys
from pathlib import Path

ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
sys.path.insert(0, str(ROOT))

import torch
from src.models.end_surv_gan import EndSurvGAN
```

## 1. Instantiate the model

The conference implementation used a latent dimension of 128 and three fully connected hidden layers with 256 units each.

```python
latent_dim = 128
n_covariates = 14  # SUPPORT example
model = EndSurvGAN(latent_dim=latent_dim, n_covariates=n_covariates, hidden_dim=256)
model
```

## 2. Generate a synthetic survival batch

```python
z = torch.randn(8, latent_dim)
x_fake, t_fake, delta_prob = model.generator(z)
print('X* shape:', x_fake.shape)
print('T* shape:', t_fake.shape)
print('Delta* probability shape:', delta_prob.shape)
print('Generated times are positive:', bool((t_fake > 0).all()))
```

## 3. Score generated observations with the critic

```python
critic_scores = model.discriminator(x_fake, t_fake, delta_prob)
critic_scores.shape
```
